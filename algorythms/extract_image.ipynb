```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
os.makedirs('output', exist_ok=True)
```

```python
PATH = 'output/'
IMAGES = './data/FashionMNIST/raw/t10k-images-idx3-ubyte'
LABELS = './data/FashionMNIST/raw/t10k-labels-idx1-ubyte'
```

```python
with open(IMAGES, 'rb') as f:
    magic = int.from_bytes(f.read(4), 'big')
    num_imgs = int.from_bytes(f.read(4), 'big')
    rows = int.from_bytes(f.read(4), 'big')
    cols = int.from_bytes(f.read(4), 'big')

    print(f'magic: {magic}, images: {num_imgs}, rows: {rows}, cols: {cols}')

    buffer = f.read(rows * cols * num_imgs)
    images = np.frombuffer(buffer, dtype=np.uint8).reshape(num_imgs, rows, cols)

with open(LABELS, 'rb') as f:
    magic = int.from_bytes(f.read(4), 'big')
    num_labels = int.from_bytes(f.read(4), 'big')

    print(f'magic: {magic}, labels: {num_labels}')

    labels = np.frombuffer(f.read(num_labels), dtype=np.uint8)


```

```python
d = {
    '0': 'zero',
    '1': 'one',
    '2': 'two',
    '3': 'three',
    '4': 'four',
    '5': 'five',
    '6': 'six',
    '7': 'seven',
    '8': 'eight',
    '9': 'nine'
}

d = {
    '0': "tshirt",
    '1': "trouser",
    '2': "pullover",
    '3': "dress",
    '4': "coat",
    '5': "sandal",
    '6': "shirt",
    '7': "sneaker",
    '8': "bag",
    '9': "boot"
}

for key in d:
    os.makedirs(os.path.join(PATH, d[key]))
```

```python
for i in range(num_imgs):
    img = images[i]
    label = labels[i]
    img = Image.fromarray(img)
    path = os.path.join(PATH, d[str(label)], f'{i}.png')
    if len(os.listdir(os.path.join(PATH, d[str(label)]))) == 100:
        continue
    img.save(path)
```
