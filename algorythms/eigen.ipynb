```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.decomposition import PCA

# Generujemy dane 2D wzdłuż skośnej linii
np.random.seed(42)
X = np.random.randn(100, 2)
X[:, 1] = 2*X[:, 0] + np.random.randn(100) * 0.5  # silna korelacja

# Tworzymy PCA i redukujemy do 1 wymiaru
pca = PCA(n_components=1)
X_reduced = pca.fit_transform(X)
X_projected = pca.inverse_transform(X_reduced)

# Rysujemy dane
plt.figure(figsize=(8,6))
plt.scatter(X[:, 0], X[:, 1], alpha=0.5, label="Oryginalne dane")
plt.scatter(X_projected[:, 0], X_projected[:, 1], alpha=0.7, label="Rzut na 1 komponent")
plt.plot([0, pca.components_[0,0]*3], [0, pca.components_[0,1]*3], color="red", linewidth=2, label="Pierwszy komponent")
plt.xlabel("Cecha 1")
plt.ylabel("Cecha 2")
plt.legend()
plt.title("PCA redukcja 2D → 1D")
plt.grid(True)
plt.show()
```
